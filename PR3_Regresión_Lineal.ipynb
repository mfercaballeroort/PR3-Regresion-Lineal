### Ejercicio 1 — Horas de estudio y nota

Queremos predecir la nota obtenida a partir de las horas de estudio.

- X = horas de estudio
- y = nota obtenida

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
horas = [1, 2, 3, 4, 5, 6]
notas = [3, 4, 5, 6, 7, 8]
```

```python
print(horas[0])
print(notas[0])
```

```python
X = np.array(horas).reshape(-1, 1)
y = np.array(notas)
```

```python
print(X)
print(y)
```

```python
print(X.shape)
print(y.shape)
```

```python
plt.scatter(X, y)

plt.xlabel("Horas de estudio")
plt.ylabel("Nota")
plt.title("Horas de estudio vs Nota")

plt.show()
```

```python
modelo = LinearRegression()
```

```python
modelo.fit(X, y)
```

```python
modelo.coef_
modelo.intercept_
```

```python
modelo.coef_
```

```python
modelo.predict([[8]])
```
