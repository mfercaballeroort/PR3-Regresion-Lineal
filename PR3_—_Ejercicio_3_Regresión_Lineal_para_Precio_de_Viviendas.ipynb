# PR3 — Ejercicio 3: Regresión Lineal para Precio de Viviendas

## Objetivo

Construir un modelo de regresión lineal que permita estimar el precio de una vivienda a partir de su superficie en metros cuadrados.

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
```

```python
superficie = [50, 60, 70, 80, 90, 100, 120]
precio = [80, 95, 110, 125, 140, 155, 185]
```

### Preparación de los datos

En este ejercicio queremos predecir el precio de una vivienda a partir de su superficie.

- **X:** superficie de la vivienda en metros cuadrados.
- **y:** precio de la vivienda, expresado en miles de dólares.

```python
X = np.array(superficie).reshape(-1, 1)
y = np.array(precio)
```

### Visualización de los datos

Utilizamos un gráfico de dispersión para observar la relación entre la superficie de la vivienda y su precio.

```python
plt.scatter(X, y)

plt.xlabel("Superficie (m²)")
plt.ylabel("Precio (miles de dólares)")
plt.title("Superficie vs Precio")

plt.show()
```

### Creación y entrenamiento del modelo

Creamos un modelo de regresión lineal para encontrar la relación entre la superficie de la vivienda y su precio.

Entrenamos el modelo utilizando `X` como variable de entrada e `y` como variable objetivo.

```python
modelo = LinearRegression()
modelo.fit(X, y)
```

### Parámetros del modelo

Consultamos la pendiente y la ordenada al origen obtenidas por el modelo para conocer la ecuación de regresión lineal.

```python
print("Pendiente:", modelo.coef_)
print("Ordenada al origen:", modelo.intercept_)
```

```python
prediccion = modelo.predict([[110]])

print("Precio predicho:", prediccion)
```

### Predicción

Utilizamos el modelo entrenado para estimar el precio de una vivienda de 110 m².

La función `predict()` recibe la superficie de la nueva vivienda y devuelve el precio estimado por el modelo.

### Resultado

El modelo estima que una vivienda de **110 m²** tendría un precio aproximado de **170 mil dólares**.
